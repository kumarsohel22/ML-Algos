# Fit method

```python
def fit(dataset):
    unique_words = set()
    
    if isinstance(dataset, (list)):
        for row in dataset:
            for word in row.split(" "):
                if len(word) < 2:
                    continue
                unique_words.add(word)
                
                #print(unique_words)
                
        unique_words = sorted(list(unique_words))
        vocab = {j:i for i,j in enumerate(unique_words)}
        
        return vocab
    else:
        print("need a list")
    
```

```python
vocab = fit(["abc def aaa prq", "lmn pqr aaaaaaa aaa abbb baaa"])
print(vocab)
```

# Transform method

```python
from collections import Counter
from scipy.sparse import csr_matrix

def transform(dataset,vocab):
    rows = []
    columns = []
    values = []
    if isinstance(dataset, (list)):
        for idx, row in enumerate(tqdm(dataset)):
            print(len(row.split(" ")))
            print(row)
            word_freq = dict(Counter(row.split()))
            #print(word_freq)
            
            for word, freq in word_freq.items():
                if len(word) <2:
                    continue
                col_index = vocab.get(word, -1)
                #print(col_index)
                if col_index !=-1:
                    # we are storing the index of the document
                    rows.append(idx)
                    # we are storing the dimensions of the word
                    columns.append(col_index)
                    # we are storing the frequency of the word
                    values.append(freq)
        return csr_matrix((values, (rows,columns)), shape=(len(dataset),len(vocab)))
    else:
        print("you need to pass list of strings")
                
```

```python
strings = ["the method of lagrange multipliers is the economists workhorse for solving optimization problems",
           "the technique is a centerpiece of economic theory but unfortunately its usually taught poorly"]
vocab = fit(strings)
print(list(vocab.keys()))
print(transform(strings, vocab).toarray())
```

```python
def tf(corpus):
    tfs = []
    for document in corpus:
        dic={}
        for word in document.split():
            if word in dic:
                dic[word]+=1
            else:
                dic[word]=1
        for word,freq in dic.items():
            #print(word,freq)
            dic[word]=freq/len(document.split())
        tfs.append(dic)
        
    return tfs


l= [
     'this is the first document',
     'this document is the second document',
     'and this is the third one',
     'is this the first document',
]
#u = fit(l)
a = tf(l)
for i in range (0,4):
    print(a[i])
```

```python
def IDF(corpus, unique_words):
    idf_dict={}
    #print(type(idf_dict))
    N=len(corpus)
    for i in unique_words:
        count=0
        for sen in corpus:
            if i in sen.split():
                count=count+1
                idf_dict[i]=(math.log((1+N)/(count+1)))+1
              
    return idf_dict
     

l= [
     'this is the first document',
     'this document is the second document',
     'and this is the third one',
     'is this the first document',
]
```
